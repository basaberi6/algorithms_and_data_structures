  Output: [1]

  ### Example 2

  Input: `lst = [5, 0, 1]`

  Input: `lst = [6, 8, 2, 3, 5, 7, 0, 1, 10]`

#### Starter Code for Question 3


```python
from typing import List

def missing_num(nums: List[int]) -> List[int] or int:
    n = len(nums)
    full_set = set(range(n + 1))
    nums_set = set(nums)
    missing = sorted(list(full_set - nums_set))
    return missing if missing else -1
```

```python
test = [6, 8, 2, 3, 5, 7, 0, 1, 10]
print(missing_num(test))
```

```python
brackets = {')': '(', ']': '[', '}': '{'}
opening_brackets = brackets.values()
closing_brackets = brackets.keys()

def is_valid_brackets(s: str) -> bool:
    stack = []
    for char in s:
        if char in opening_brackets:
            stack.append(char)
        elif char in closing_brackets:
            if len(stack) == 0 or stack.pop() != brackets[char]:
                return False
        else:
            raise ValueError(f"Invalid character: '{char}'")
    return len(stack) == 0
```
